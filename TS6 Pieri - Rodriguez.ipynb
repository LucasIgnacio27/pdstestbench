## Introducción

En este trabajo práctico analizamos en el dominio de la frecuencia compleja ($s$) distintos filtros de segundo orden armados con redes $RLC$ pasivas. Para todos los circuitos usamos los valores normalizados $\omega_0 = 1\ \text{rad/s}$ y $R = 1\ \Omega$, con un factor de calidad $Q = \frac{\sqrt{2}}{2}$, que es el de un filtro Butterworth (respuesta plana).

Lo que hacemos en cada circuito es lo siguiente:

1. **Función transferencia:** calculamos $H(s) = \frac{V_2(s)}{V_1(s)}$ con la salida en vacío ($I_2 = 0$).
2. **Bode:** graficamos módulo y fase de $H(j\omega)$ junto con sus asíntotas, y ubicamos los polos y ceros en el plano $s$.
3. **Tipo de filtro y ancho de banda:** determinamos si cada red es pasa-bajos, pasa-altos o pasa-banda, y calculamos el ancho de banda con el criterio de $-3\ \text{dB}$.
4. **Circuito Bonus:** estudiamos la transferencia de la red de segundo orden de estructura puenteada y cómo cambia según los parámetros $a$ y $b$.

## Circuito a 

### Modelo Eléctrico y Análisis Teórico
Este circuito consiste en una inductancia $L$, una resistencia $R$ y un capacitor $C$ conectados en serie, alimentados por una tensión de entrada $V_1(s)$ y tomando la salida $V_2(s)$ sobre bornes del capacitor $C$ en condición de vacío ($I_2 = 0$).

#### Análisis de extremos en frecuencia:
* **En corriente continua ($\omega \to 0$ / $s \to 0$):** El capacitor se comporta como un circuito abierto ($Z_C = \frac{1}{sC} \to \infty$). No circula corriente, por lo que no hay caída en $R$ ni en $L$, resultando $V_2(0) = V_1(0)$ (ganancia unitaria o $0 \text{ dB}$).
* **En alta frecuencia ($\omega \to \infty$ / $s \to \infty$):** El capacitor actúa como un cortocircuito a masa ($Z_C \to 0$), por lo que $V_2(\infty) = 0 \text{ V}$ ($-\infty \text{ dB}$).

**Conclusión preliminar:** El circuito atenúa las altas frecuencias y deja pasar las bajas, comportándose como un **filtro pasa-bajos de segundo orden**.

---

### Deducción Analítica de la Función de Transferencia $H_a(s)$

Aplicando el divisor de tensión, ($I_2 = 0$):

$$H_a(s) = \frac{V_2(s)}{V_1(s)} = \frac{Z_C(s)}{Z_L(s) + Z_R(s) + Z_C(s)} = \frac{\frac{1}{sC}}{sL + R + \frac{1}{sC}}$$

Multiplicando numerador y denominador por $\frac{s}{L}$:

$$H_a(s) = \frac{\frac{1}{LC}}{s^2 + \frac{R}{L}s + \frac{1}{LC}}$$

#### Determinación de los valores de $L$ y $C$:
Comparando con la forma canónica del denominador de segundo orden $D(s) = s^2 + \frac{\omega_0}{Q}s + \omega_0^2$:
1. $\omega_0^2 = \frac{1}{LC} \implies LC = 1 \quad (\text{ya que } \omega_0 = 1 \text{ rad/s})$
2. $\frac{\omega_0}{Q} = \frac{R}{L} \implies \frac{1}{\frac{\sqrt{2}}{2}} = \frac{1}{L} \implies \sqrt{2} = \frac{1}{L} \implies L = \frac{1}{\sqrt{2}} = \frac{\sqrt{2}}{2} \text{ H} \approx 0,7071 \text{ H}$
3. $C = \frac{1}{\omega_0^2 L} = \sqrt{2} \text{ F} \approx 1,4142 \text{ F}$

Reemplazando estos valores en la expresión de transferencia se obtiene la función normalizada:

$$H_a(s) = \frac{1}{s^2 + \sqrt{2}s + 1}$$

---

### Diagrama de Polos y Ceros, Estabilidad
* **Ceros:** No posee ceros finitos (presenta $2$ ceros en el infinito, $s \to \infty$).
* **Polos:** Obtenidos de $s^2 + \sqrt{2}s + 1 = 0$:
  $$s_{1,2} = -\frac{\sqrt{2}}{2} \pm j\sqrt{1 - \left(\frac{\sqrt{2}}{2}\right)^2} = -\frac{\sqrt{2}}{2} \pm j\frac{\sqrt{2}}{2} \approx -0,7071 \pm j0,7071$$
* **Estabilidad:** Al poseer ambos polos en el semiplano izquierdo ($\text{Re}(s) = -0,7071 < 0$), el sistema es **asintóticamente estable**.

---

### Respuesta de Magnitud y Fase 

#### Magnitud $|H_a(j\omega)|$:
$$|H_a(j\omega)| = \frac{1}{|(1 - \omega^2) + j\sqrt{2}\omega|} = \frac{1}{\sqrt{(1 - \omega^2)^2 + 2\omega^2}} = \frac{1}{\sqrt{\omega^4 + 1}}$$

* **Banda de paso ($\omega \ll 1$):** $|H_a(j\omega)| \approx 1 \implies 0 \text{ dB}$ (asíntota horizontal).
* **Frecuencia de corte ($\omega = \omega_0 = 1 \text{ rad/s}$):** $|H_a(j1)| = \frac{1}{\sqrt{2}} \approx 0,7071 \implies -3,01 \text{ dB}$.
* **Banda de atenuación ($\omega \gg 1$):** $|H_a(j\omega)| \approx \frac{1}{\omega^2} \implies -40 \text{ dB/década}$ (asíntota de alta frecuencia con pendiente de $-2$).

#### Fase $\phi_a(\omega)$:
$$\phi_a(\omega) = -\arctan\left(\frac{\sqrt{2}\omega}{1 - \omega^2}\right)$$

* $\omega \to 0 \implies \phi_a \to 0^\circ$
* $\omega = \omega_0 = 1 \text{ rad/s} \implies \phi_a = -90^\circ$
* $\omega \to \infty \implies \phi_a \to -180^\circ$

---

### Tipo de Filtro y Ancho de Banda
* **Tipo de filtro:** Pasa-bajos de $2.^\text{do}$ orden tipo **Butterworth** ($Q = \frac{\sqrt{2}}{2}$, condición de máxima planicidad en la banda de paso).
* **Ancho de banda ($BW$):** $BW = \omega_c = 1 \text{ rad/s}$.

## Circuito b

Para este circuito, analizamos la rama de entrada formada por la combinación en **paralelo** de la resistencia $R$ y el capacitor $C$, la cual se encuentra conectada en serie con la inductancia $L$ derivada a masa.

La impedancia equivalente del conjunto $R \parallel C$ está dada por:

$$Z_{RC}(s) = \frac{R}{1 + RCs}$$

A partir de esta configuración, procedemos a deducir la función de transferencia $H(s) = \frac{V_o(s)}{V_i(s)}$ mediante el divisor de tensión entre $Z_{RC}(s)$ y la impedancia del inductor $Z_L(s) = sL$:

$$H(s) = \frac{Z_{RC}(s)}{Z_{RC}(s) + Z_L(s)}$$

#### Análisis de extremos en frecuencia:
* **En corriente continua ($\omega \to 0$ / $s \to 0$):** La inductancia $L$ se comporta como un cortocircuito a masa ($Z_L = sL \to 0$), cancelando por completo la señal de salida ($V_2(0) = 0 \text{ V} \implies -\infty \text{ dB}$).
* **En alta frecuencia ($\omega \to \infty$ / $s \to \infty$):** El capacitor $C$ en el bloque paralelo actúa como un cortocircuito ($Z_{RC} \to 0$) y la inductancia $L$ como un circuito abierto ($Z_L \to \infty$), permitiendo el paso íntegro de la tensión de entrada a la salida ($V_2(\infty) = V_1(\infty) \implies 0 \text{ dB}$).

**Conclusión preliminar:** El circuito bloquea la componente continua y las bajas frecuencias, dejando pasar las altas frecuencias, por lo que se comporta como un **filtro pasa-altos de segundo orden**.

---

### Deducción Analítica de la Función de Transferencia $H_b(s)$

1. **Impedancia equivalente del bloque $R \parallel C$:**
   $$Z_{RC}(s) = \frac{R \cdot \frac{1}{sC}}{R + \frac{1}{sC}} = \frac{R}{1 + RCs}$$

2. **Divisor de tensión ($I_2 = 0$):**
   $$H_b(s) = \frac{V_2(s)}{V_1(s)} = \frac{Z_L(s)}{Z_{RC}(s) + Z_L(s)} = \frac{sL}{\frac{R}{1 + RCs} + sL}$$

3. **Acondicionamiento algebraico:**
   Multiplicando numerador y denominador por $(1 + RCs)$:
   $$H_b(s) = \frac{sL(1 + RCs)}{R + sL(1 + RCs)} = \frac{s^2 RLC + sL}{s^2 RLC + sL + R}$$

   Dividiendo todo por $RLC$:
   $$H_b(s) = \frac{s \left(s + \frac{1}{RC}\right)}{s^2 + \frac{1}{RC}s + \frac{1}{LC}}$$

4. **Determinación de los componentes normalizados:**
   Para conservar el denominador canónico Butterworth $s^2 + \sqrt{2}s + 1$ (con $\omega_0 = 1 \text{ rad/s}$, $R = 1 \ \Omega$ y $Q = \frac{\sqrt{2}}{2}$):
   * $\frac{1}{RC} = \sqrt{2} \implies C = \frac{1}{\sqrt{2}R} = \frac{\sqrt{2}}{2} \text{ F} \approx 0,7071 \text{ F}$
   * $\frac{1}{LC} = 1 \implies L = \frac{1}{C} = \sqrt{2} \text{ H} \approx 1,4142 \text{ H}$

   Reemplazando estos parámetros en la función de transferencia:

   $$H_b(s) = \frac{s(s + \sqrt{2})}{s^2 + \sqrt{2}s + 1}$$

---

### Diagrama de Polos y Ceros, Estabilidad
* **Ceros:** 
  * $s_1 = 0$ (un cero en el origen).
  * $s_2 = -\sqrt{2} \approx -1,4142$ (un cero en el eje real negativo).
* **Polos:** Idénticos a los del circuito **a**:
  $$s_{1,2} = -\frac{\sqrt{2}}{2} \pm j\frac{\sqrt{2}}{2} \approx -0,7071 \pm j0,7071$$
* **Estabilidad:** **Asintóticamente estable** ($\text{Re}(s) = -0,7071 < 0$).

---

###  Respuesta de Magnitud y Fase 

#### Magnitud $|H_b(j\omega)|$:
$$|H_b(j\omega)| = \frac{|j\omega(j\omega + \sqrt{2})|}{|(1 - \omega^2) + j\sqrt{2}\omega|} = \frac{\omega \sqrt{\omega^2 + 2}}{\sqrt{(1 - \omega^2)^2 + 2\omega^2}} = \frac{\sqrt{\omega^4 + 2\omega^2}}{\sqrt{\omega^4 + 1}}$$

* **Banda de atenuación ($\omega \ll 1$):** $|H_b(j\omega)| \approx \sqrt{2}\omega \implies +20 \text{ dB/década}$ (asíntota de baja frecuencia determinada por el cero en el origen).
* **Banda de paso ($\omega \gg 1$):** $|H_b(j\omega)| \approx 1 \implies 0 \text{ dB}$ (asíntota horizontal de alta frecuencia).
* **Frecuencia de corte ($-\text{3 dB}$):**
  Igualando $|H_b(j\omega_c)|^2 = \frac{1}{2}$:
  $$\frac{\omega_c^4 + 2\omega_c^2}{\omega_c^4 + 1} = \frac{1}{2} \implies 2\omega_c^4 + 4\omega_c^2 = \omega_c^4 + 1 \implies \omega_c^4 + 4\omega_c^2 - 1 = 0$$
  Resolviendo la cuadrática para $\omega_c^2$:
  $$\omega_c^2 = \frac{-4 + \sqrt{16 + 4}}{2} = \sqrt{5} - 2 \approx 0,2361 \implies \omega_c = \sqrt{\sqrt{5} - 2} \approx 0,4859 \text{ rad/s}$$

#### Fase $\phi_b(\omega)$:
$$\phi_b(\omega) = 90^\circ + \arctan\left(\frac{\omega}{\sqrt{2}}\right) - \arctan\left(\frac{\sqrt{2}\omega}{1 - \omega^2}\right)$$

* $\omega \to 0 \implies \phi_b \to +90^\circ$
* $\omega \to \infty \implies \phi_b \to 0^\circ$

---

### Tipo de Filtro y Ancho de Banda
* **Tipo de filtro:** Pasa-altos de $2.^\text{do}$ orden con cero de transmisión adicional en $s = -\sqrt{2}$.
* **Ancho de banda ($BW$):** $BW \to \infty$ en el rango de paso $[\omega_c, \infty)$ con $\omega_c \approx 0,4859 \text{ rad/s}$.

## Circuito c 

###  Modelo Eléctrico y Análisis Teórico
En el circuito **c**, la señal de entrada $V_1(s)$ pasa por una resistencia $R$ en serie y alimenta un tanque resonante paralelo compuesto por un inductor $L$ y un capacitor $C$ conectados a masa. La señal de salida $V_2(s)$ se toma directamente sobre el tanque $L \parallel C$ en condición de vacío ($I_2 = 0$).

#### Análisis de extremos en frecuencia:
* **En corriente continua ($\omega \to 0$ / $s \to 0$):** El inductor actúa como un cortocircuito a masa ($Z_L = sL \to 0$), por lo que $V_2(0) = 0 \text{ V}$ ($-\infty \text{ dB}$).
* **En alta frecuencia ($\omega \to \infty$ / $s \to \infty$):** El capacitor actúa como un cortocircuito a masa ($Z_C = \frac{1}{sC} \to 0$), por lo que $V_2(\infty) = 0 \text{ V}$ ($-\infty \text{ dB}$).
* **En la frecuencia de resonancia ($\omega = \omega_0 = \frac{1}{\sqrt{LC}}$):** La impedancia del tanque paralelo entra en resonancia ($Z_p \to \infty$), comportándose como un circuito abierto. La corriente que circula por $R$ es nula, logrando la máxima transferencia $V_2(\omega_0) = V_1(\omega_0)$ (ganancia unitaria o $0 \text{ dB}$).

**Conclusión preliminar:** El circuito atenúa las frecuencias muy bajas y muy altas, dejando pasar únicamente una banda alrededor de $\omega_0$, comportándose como un **filtro pasa-banda de segundo orden**.

---

###  Deducción Analítica de la Función de Transferencia $H_c(s)$

1. **Impedancia equivalente del tanque paralelo $L \parallel C$:**
   $$Z_p(s) = \frac{sL \cdot \frac{1}{sC}}{sL + \frac{1}{sC}} = \frac{sL}{s^2 LC + 1}$$

2. **Divisor de tensión ($I_2 = 0$):**
   $$H_c(s) = \frac{V_2(s)}{V_1(s)} = \frac{Z_p(s)}{R + Z_p(s)} = \frac{\frac{sL}{s^2 LC + 1}}{R + \frac{sL}{s^2 LC + 1}}$$

3. **Acondicionamiento algebraico:**
   Multiplicando numerador y denominador por $(s^2 LC + 1)$:
   $$H_c(s) = \frac{sL}{R(s^2 LC + 1) + sL} = \frac{sL}{s^2 RLC + sL + R}$$

   Dividiendo todo por $RLC$:
   $$H_c(s) = \frac{\frac{1}{RC}s}{s^2 + \frac{1}{RC}s + \frac{1}{LC}}$$

4. **Sustitución de componentes normalizados:**
   Con $\omega_0 = 1 \text{ rad/s}$, $R = 1 \ \Omega$ y $Q = \frac{\sqrt{2}}{2}$, los componentes del tanque resultan $C = \frac{\sqrt{2}}{2} \text{ F} \approx 0,7071 \text{ F}$ y $L = \sqrt{2} \text{ H} \approx 1,4142 \text{ H}$:
   $$\frac{1}{RC} = \frac{1}{1 \cdot \frac{\sqrt{2}}{2}} = \sqrt{2}$$
   $$\frac{1}{LC} = \frac{1}{\sqrt{2} \cdot \frac{\sqrt{2}}{2}} = 1$$

   Sustituyendo obtenemos la función de transferencia normalizada:

   $$H_c(s) = \frac{\sqrt{2}s}{s^2 + \sqrt{2}s + 1}$$

---

### Diagrama de Polos y Ceros, Estabilidad
* **Ceros:** $s_1 = 0$ (un cero en el origen) y un cero en el infinito.
* **Polos:** Mismos polos característicos del denominador cuadrático:
  $$s_{1,2} = -\frac{\sqrt{2}}{2} \pm j\frac{\sqrt{2}}{2} \approx -0,7071 \pm j0,7071$$
* **Estabilidad:** **Asintóticamente estable** ($\text{Re}(s) = -0,7071 < 0$).

---

### Respuesta de Magnitud, Fase y Diagrama de Bode

#### Magnitud $|H_c(j\omega)|$:
$$|H_c(j\omega)| = \frac{\sqrt{2}\omega}{|(1 - \omega^2) + j\sqrt{2}\omega|} = \frac{\sqrt{2}\omega}{\sqrt{(1 - \omega^2)^2 + 2\omega^2}} = \frac{\sqrt{2}\omega}{\sqrt{\omega^4 + 1}}$$

* **Baja frecuencia ($\omega \ll 1$):** $|H_c(j\omega)| \approx \sqrt{2}\omega \implies +20 \text{ dB/década}$ (asíntota por el cero en el origen).
* **Frecuencia central ($\omega = \omega_0 = 1 \text{ rad/s}$):** $|H_c(j1)| = \frac{\sqrt{2}}{\sqrt{2}} = 1 \implies 0 \text{ dB}$ (ganancia máxima).
* **Alta frecuencia ($\omega \gg 1$):** $|H_c(j\omega)| \approx \frac{\sqrt{2}}{\omega} \implies -20 \text{ dB/década}$ (asíntota de caída de primer orden).

#### Frecuencias de corte a $-3 \text{ dB}$ y Ancho de Banda:
Igualando $|H_c(j\omega)|^2 = \frac{1}{2}$:
$$\frac{2\omega^2}{\omega^4 + 1} = \frac{1}{2} \implies 4\omega^2 = \omega^4 + 1 \implies \omega^4 - 4\omega^2 + 1 = 0$$

Resolviendo la cuadrática para $\omega^2$:
$$\omega^2 = \frac{4 \pm \sqrt{16 - 4}}{2} = 2 \pm \sqrt{3}$$

* **Frecuencia de corte inferior ($\omega_1$):** $\omega_1 = \sqrt{2 - \sqrt{3}} \approx 0,5176 \text{ rad/s}$
* **Frecuencia de corte superior ($\omega_2$):** $\omega_2 = \sqrt{2 + \sqrt{3}} \approx 1,9318 \text{ rad/s}$
* **Propiedad de simetría geométrica:** $\omega_1 \cdot \omega_2 = \sqrt{(2 - \sqrt{3})(2 + \sqrt{3})} = \sqrt{4 - 3} = 1 = \omega_0$
* **Ancho de banda ($BW$):** 
  $$BW = \omega_2 - \omega_1 = \frac{\omega_0}{Q} = \frac{1}{\frac{\sqrt{2}}{2}} = \sqrt{2} \approx 1,4142 \text{ rad/s}$$

#### Fase $\phi_c(\omega)$:
$$\phi_c(\omega) = 90^\circ - \arctan\left(\frac{\sqrt{2}\omega}{1 - \omega^2}\right)$$

* $\omega \to 0 \implies \phi_c \to +90^\circ$
* $\omega = \omega_0 = 1 \text{ rad/s} \implies \phi_c = 0^\circ$
* $\omega \to \infty \implies \phi_c \to -90^\circ$

---

### Tipo de Filtro y Ancho de Banda
* **Tipo de filtro:** Pasa-banda de $2.^\text{do}$ orden.
* **Ancho de banda ($BW$):** $BW = \sqrt{2} \approx 1,4142 \text{ rad/s}$ en el intervalo de paso $[0,5176; 1,9318] \text{ rad/s}$.

## Conclusión General del Trabajo

A través del análisis analítico, numérico y gráfico desarrollado para las tres configuraciones $RLC$, se obtuvieron las siguientes conclusiones principales:

* **Circuito a:**
  Actúa como un filtro pasa-bajos de segundo orden con alineación **Butterworth** ($Q = \sqrt{2}/2 \approx 0,7071$). A frecuencias bajas ($\omega \to 0$), el capacitor se comporta como un circuito abierto permitiendo la transferencia total de la señal ($0 \text{ dB}$). A altas frecuencias, deriva la corriente a masa atenuando la salida con una pendiente asintótica de $-40 \text{ dB/década}$. No posee ceros finitos y su ancho de banda es $BW = 1 \text{ rad/s}$.

* **Circuito b:**
  Funciona como un filtro pasa-altos de segundo orden. En continua ($\omega = 0$), el inductor cortocircuita la salida a masa ($0 \text{ V}$). A frecuencias elevadas ($\omega \to \infty$), la combinación $RC$ no presenta impedancia apreciable y la salida toma todo el valor de entrada ($0 \text{ dB}$). La presencia de un cero en el origen ($s = 0$) y un cero en el eje real ($s = -\sqrt{2}$) acelera la transición de fase y fija la frecuencia de corte a $-3 \text{ dB}$ en $\omega_c \approx 0,4859 \text{ rad/s}$ ($BW \to \infty$).

* **Circuito c:**
  Se comporta como un filtro pasa-banda de segundo orden. En los extremos de frecuencia ($\omega \to 0$ y $\omega \to \infty$), la impedancia del tanque paralelo tiende a cero, atenuando la salida. En la frecuencia de resonancia ($\omega_0 = 1 \text{ rad/s}$), el tanque entra en resonancia ($Z_p \to \infty$) logrando la ganancia máxima de $0 \text{ dB}$ y fase $0^\circ$. Presenta un cero en el origen ($s = 0$) y un ancho de banda de $BW = \sqrt{2} \approx 1,4142 \text{ rad/s}$.

---

### Síntesis 
Los tres circuitos comparten exactamente la **misma ecuación característica en el denominador** ($D(s) = s^2 + \sqrt{2}s + 1$), lo que se traduce en una distribución idéntica de polos en el plano complejo ($s_{1,2} = -0,7071 \pm j0,7071$). Esto garantiza que las tres redes poseen el mismo comportamiento dinámico natural y son **asintóticamente estables** (polos estrictamente en el semiplano izquierdo).

La diferencia en la respuesta frecuencial (pasa-bajos, pasa-altos o pasa-banda) no depende del denominador sino exclusivamente del **nodo donde se extrae la tensión de salida $V_o(s)$**, lo cual define la cantidad y posición de los **ceros** en el numerador de la función de transferencia.

## Circuito Bonus

### Parámetros de Diseño:
* Frecuencia deseada: $f_0 = 10\text{ kHz} \implies \omega_0' = 2\pi f_0 = 20000\pi \approx 62831.85\text{ rad/s}$
* Resistencia deseada: $R' = 50\,\Omega$

Los parámetros $a$ y $b$ son adimensionales, por lo que el escalado no los modifica.

---

### Función de transferencia (verificación)

$$H(s)=\frac{V_2}{V_1}=\frac{a\,LC\,s^2+b}{LC\,s^2+\frac{L}{R}s+1}$$

El denominador no depende de $a$ ni de $b$, así que $\omega_0 = \frac{1}{\sqrt{LC}}$ y $Q = R\sqrt{\frac{C}{L}} = 0.707$ se mantienen para cualquier $a$ y $b$.

---

### Componentes normalizados (con $\omega_0 = 1\text{ rad/s}$):
$$R_{norm} = 1\,\Omega \quad \vert{} \quad L_{norm} = \sqrt{2}\text{ H} \approx 1.4142\text{ H} \quad \vert{} \quad C_{norm} = \frac{\sqrt{2}}{2}\text{ F} \approx 0.7071\text{ F}$$

---

### Cálculo de Componentes Desescalados

* **Resistencia ($R'$):**
  $$R' = 50\,\Omega$$

* **Capacitor serie ($aC'$):**
  $$aC' = a \cdot \frac{\sqrt{2}}{2} \cdot \frac{1}{50 \cdot 20000\pi} \approx a \cdot 225.08\text{ nF}$$

* **Inductor serie ($L'/b$):**
  $$\frac{L'}{b} = \frac{1}{b} \cdot \sqrt{2} \cdot \frac{50}{20000\pi} \approx \frac{1.1254\text{ mH}}{b}$$

* **Capacitor a masa ($(1-a)C'$):**
  $$(1-a)C' = (1-a) \cdot \frac{\sqrt{2}}{2} \cdot \frac{1}{50 \cdot 20000\pi} \approx (1-a) \cdot 225.08\text{ nF}$$

* **Inductor a masa ($L'/(1-b)$):**
  $$\frac{L'}{1-b} = \frac{1}{1-b} \cdot \sqrt{2} \cdot \frac{50}{20000\pi} \approx \frac{1.1254\text{ mH}}{1-b}$$

### Simulación en LTspice 

Con los valores desescalados para $f_0 = 10\ \text{kHz}$ y $R = 50\ \Omega$, la función transferencia del circuito es

$$H(s)=\frac{a\,s^2+b\,\omega_0^2}{s^2+\frac{\omega_0}{Q}s+\omega_0^2}$$

Como el denominador no depende de $a$ ni de $b$, los polos quedan fijos. Los parámetros $a$ y $b$ solo modifican los ceros de transmisión, y con eso cambia el tipo de filtro. Se simularon dos casos.

Es un pasa-bajos de segundo orden con ganancia unitaria en continua ($H(0)=b=1$) y caída de $-40\ \text{dB/déc}$ a partir de $f_0$. # Conclusión

El circuito Bonus muestra que se puede reorganizar cómo se reparten la capacidad y la inductancia entre la rama serie y la rama a masa sin alterar la dinámica propia del sistema. Los parámetros $a$ y $b$ solo dividen cada elemento en dos partes complementarias, por lo que la cantidad total de $C$ y de $L$ que ve el circuito no cambia. Como la frecuencia característica y los polos dependen únicamente de esos totales y de $R$, el desescalado a $f_0 = 10\text{ kHz}$ y $R = 50\,\Omega$ conserva $\omega_0$ y $Q = 0.707$ para cualquier elección de $a$ y $b$.

Lo que cambia es cómo la señal llega a la salida: $a$ y $b$ determinan los ceros de la transferencia, es decir, qué frecuencias se transmiten y cuáles se bloquean. De este modo, una misma "base" de polos permite obtener un pasabajos, un pasaaltos o un notch con solo redistribuir los componentes.
